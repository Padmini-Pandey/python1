Maximum of 2 numbers

```python
def max(a,b):
  if(a>b):
    return a
  else:
    return b
m=int(input("enter a number "))
n=int(input("enter a number "))
print("maximum number is ",max(m,n))
```

factorial

```python
def fact(n):
  f=1
  for i in range(n,0,-1):
    f=f*i
  return f
a=int(input("enter an integer "))
print("Factorial of a number ",fact(a))
```

Celsius to Fahrenheit

```python
def CtoF(ce):
  f=(ce*(9/5))+32
  return f
c=float(input("enter temperature in celsius "))
print("temperature in Fahrenheit ",CtoF(c))
```

list mutation

```python
def remove_last(lst):
  l=len(lst)
  temp=[]
  for i in range(0,l-1):
      temp.append(lst[i])
  return temp
li=input("enter elements separated by space ").split()
print("after mutation ",remove_last(li))
```

string reassignment

```python
def change_string(s):
  s="X"+s[1:]
  return s
a=input("enter a string ")
print("changed string ",change_string(a))
```

Dictionary mutation and rebinding

```python
def add_entry(d):
  d["number"]=123456
d={"name":"Padmini"}
add_entry(d)
print("after add entry ",d)
#rebinding
def reassign_dict(d):
  d["name"]="Rahul"
  print("updated dictionary",d)
reassign_dict(d)
```
